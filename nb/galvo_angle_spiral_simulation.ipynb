# spiralness paramter B simulation

```python
# spiralness paramter B simulation
from ipywidgets import interact, FloatSlider

def plot_spiral(B=0.25):
    A = 1.0
    modulations = 5
    theta_max = modulations * 2 * np.pi
    theta = np.linspace(0, theta_max, 150)
    r = (A/2) + (A/2) * np.sin(B * theta)
    min_amp_index = np.argmin(r)
    max_amp_index = np.argmax(r)
    index_before, index_after = (min_amp_index, max_amp_index) if min_amp_index < max_amp_index else (max_amp_index, min_amp_index)
    r = r[index_before: index_after]
    theta = theta[index_before: index_after ]
    # r = np.concatenate((r[::-1], -r))
    # theta = np.concatenate((theta[::-1], theta))

    if np.abs(r[0]) < np.abs(r[-1]):
        r = np.concatenate((r[::-1], -r, ))  # Create a continuous spiral in/out effect
        theta = np.concatenate((theta[::-1], theta, ))  # Offset theta for the second half
    else:
        r = np.concatenate((r, -r[::-1], ))  # Create a continuous spiral in/out effect
        theta = np.concatenate((theta, theta[::-1], ))  # Offset theta for the second half

    x = r * np.cos(theta)
    y = r * np.sin(theta)
    fig, ax = plt.subplots(1, 2, figsize=(16, 8))
    ax[0].plot(theta, r, color='blue')
    ax[0].set_title('Radial Behavior (Theta vs Radius)')
    ax[0].set_xlabel('Theta (radians)')
    ax[0].set_ylabel('Radius')
    ax[0].grid(True)
    ax[1].plot(x, y, color='red')
    ax[1].set_title('Spiral In/Out Pattern')
    ax[1].set_xlabel('X')
    ax[1].set_ylabel('Y')
    ax[1].axis('equal')
    ax[1].grid(True)
    perimeter = plt.Circle((0, 0), A, fill=False, color='green', linestyle='--', alpha=0.7)
    ax[1].add_patch(perimeter)
    plt.suptitle(f"Continuous Spiral In/Out with Circular Perimeter (B={B:.3f})", fontsize=16)
    plt.tight_layout()
    plt.show()

interact(plot_spiral, B=FloatSlider(value=0.25, min=0.05, max=1.0, step=0.01))
```

```python
from ipywidgets import interact, IntSlider

def spiral_plot(theta_index=0):
    A = 1.0
    B = 0.2
    modulations = 5
    theta = np.linspace(0, modulations * 2 * np.pi, 150)
    r = (A/2) + (A/2) * np.sin(B * theta)
    min_amp_index = np.argmin(r)
    max_amp_index = np.argmax(r)
    index_before, index_after = (min_amp_index, max_amp_index) if min_amp_index < max_amp_index else (max_amp_index, min_amp_index)
    r = r[index_before: index_after]
    theta = theta[index_before: index_after ]
    rr = np.array([1]*30)
    # path for full circle
    theta_full = np.linspace(0, modulations*np.pi/2, 30)
    if np.abs(r[0]) < np.abs(r[-1]):
        r = np.concatenate((r[::-1], -r, ))
        theta = np.concatenate((theta[::-1], theta, ))
    else:
        r = np.concatenate((rr,  r, -r[::-1], -rr[::-1]))
        theta = np.concatenate((theta_full, theta, theta[::-1], theta_full[::-1]))
    x = r * np.cos(theta)
    y = r * np.sin(theta)
    fig, ax = plt.subplots(1, 2, figsize=(16, 8))
    ax[0].plot(theta[:theta_index], r[:theta_index], color='blue')
    ax[0].set_title('Radial Behavior (Theta vs Radius)')
    ax[0].set_xlabel('Theta (radians)')
    ax[0].set_ylabel('Radius')
    ax[0].grid(True)
    ax[1].plot(x[:theta_index], y[:theta_index], color='red', alpha=0.7)
    # add marker for current point
    ax[1].scatter(x[theta_index], y[theta_index], color='black', s=50, zorder=5)
    ax[1].set_title('Spiral In/Out Pattern')
    ax[1].set_xlabel('X')
    ax[1].set_ylabel('Y')
    ax[1].axis('equal')
    ax[1].grid(True)
    perimeter = plt.Circle((0, 0), A, fill=False, color='green', linestyle='--', alpha=0.7)
    ax[1].add_patch(perimeter)
    plt.suptitle(f"Continuous Spiral In/Out with Circular Perimeter (index={theta_index:.2f})", fontsize=16)
    plt.tight_layout()
    plt.show()

interact(spiral_plot, theta_index=IntSlider(value=0, min=0, max=209, step=1))
```
